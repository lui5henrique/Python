```python
import pandas as pd
```

```python
empenhos_emitidos = pd.read_excel('empenhos_emitidos.xlsx', header = 2)
display(empenhos_emitidos.head(3))
empenhos_emitidos.info()
```

```python
# excluindo os espaçõs desnecessários no começo e final das colunas de texto
empenhos_emitidos = empenhos_emitidos.apply(lambda x: x.str.strip() if x.dtype == "object" else x)
```

```python
excluir_colunas = ['NE - PTRES']
empenhos_emitidos = empenhos_emitidos.drop(columns = excluir_colunas)
display(empenhos_emitidos)
```

```python
renomear_colunas = {
    'NE CCor': 'empenho',
    'NE CCor - Favorecido': 'doc_favorecido',
    'Unnamed: 2': 'nome_favorecido',
    'NE - Natureza Despesa': 'cod_natu_despesa',
    'Unnamed: 4': 'natureza_despesa',
    'NE - PI': 'cod_plano_interno',
    'Unnamed: 6': 'nome_plano_interno',
    'Ano Lançamento': 'ano_lancamento',
    'Saldo - Moeda Origem (Conta Contábil)': 'saldo_moeda'
}
empenhos_emitidos = empenhos_emitidos.rename(columns=renomear_colunas)
display(empenhos_emitidos)
empenhos_emitidos.info()
```

```python
empenhos_emitidos.to_excel('empenhos_emitidos_tratada.xlsx', sheet_name='empenhos_emitidos', index=False)
```
