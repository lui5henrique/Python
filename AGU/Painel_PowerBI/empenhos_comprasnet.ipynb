```python
import pandas as pd
```

```python
empenhos_comprasnet = pd.read_excel('empenhos_comprasnet.xlsx', header=1)
display(empenhos_comprasnet.head(3))
empenhos_comprasnet.info()
```

```python
# excluindo os espaçõs desnecessários no começo e final das colunas de texto
empenhos_comprasnet = empenhos_comprasnet.apply(lambda x: x.str.strip() if x.dtype == "object" else x)
```

```python
excluir_colunas = [
    'Atualizado Em'
]
empenhos_comprasnet = empenhos_comprasnet.drop(columns= excluir_colunas)

display(empenhos_comprasnet.head(3))
empenhos_comprasnet.info()
```

```python
renomear_colunas = {
    'Número do instrumento': 'num_instrumento',
    'Fornecedor': 'favorecido',
    'Objeto': 'objeto',
    'Número Empenho': 'num_empenho',
    'Plano Interno': 'plano_interno',
    'Natureza Despesa': 'natureza_despesa',
    'Empenhado': 'empenhado',
    'a Liquidar': 'a_liquidar',
    'Liquidado': 'liquidado',
    'Pago': 'pago',
    'RP Inscrito': 'RP_inscrito',
    'RP a Liquidar': 'RP_a_liquidar',
    'RP Liquidado': 'RP_liquidado',
    'RP Pago': 'RP_pago'
}
empenhos_comprasnet = empenhos_comprasnet.rename(columns= renomear_colunas)

display(empenhos_comprasnet.head(5))
empenhos_comprasnet.info()
```

```python
empenhos_comprasnet = empenhos_comprasnet[empenhos_comprasnet['num_empenho'].str.startswith(('2025', '2026'))]

display(empenhos_comprasnet)
empenhos_comprasnet.info()
```
